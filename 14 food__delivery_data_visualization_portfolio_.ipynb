```python
# DAY 14: FOOD DELIVERY DATA VISUALIZATION PORTFOLIO

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

# Plotting settings
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported successfully.")

# ============================================================
# 2. LOAD THE DATASET
# ============================================================

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("File:", file_name)
print("Shape:", df.shape)

# ============================================================
# 2. LOAD THE DATASET
# ============================================================

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("File:", file_name)
print("Shape:", df.shape)

# ============================================================
# 3. INITIAL DATASET INSPECTION
# ============================================================

print("First 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe(include="all").T)

# ============================================================
# 4. PREPARE DATA
# ============================================================

# Convert date column
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Display numerical and categorical columns
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include="object").columns.tolist()

print("Numerical columns:")
print(numeric_columns)

print("\nCategorical columns:")
print(categorical_columns)

# Check the date range if Date exists
if "Date" in df.columns:
    print("\nDate range:")
    print(df["Date"].min(), "to", df["Date"].max())

# ============================================================
# 5. HISTOGRAM - ORDER DISTRIBUTION
# ============================================================

if "Orders" in df.columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        data=df,
        x="Orders",
        bins=20,
        kde=True
    )

    plt.title("Distribution of Food Delivery Orders", fontsize=16)
    plt.xlabel("Number of Orders")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()
# ============================================================
# 6. HISTOGRAM - REVENUE DISTRIBUTION
# ============================================================

if "Revenue" in df.columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        data=df,
        x="Revenue",
        bins=20,
        kde=True
    )

    plt.title("Distribution of Food Delivery Revenue", fontsize=16)
    plt.xlabel("Revenue")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()

# ============================================================
# 7. COUNT PLOT - RECORDS BY CITY
# ============================================================

if "City" in df.columns:

    plt.figure(figsize=(10, 6))

    sns.countplot(
        data=df,
        x="City",
        order=df["City"].value_counts().index
    )

    plt.title("Number of Records by City", fontsize=16)
    plt.xlabel("City")
    plt.ylabel("Number of Records")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ============================================================
# 8. BAR CHART - REVENUE BY CITY
# ============================================================

if {"City", "Revenue"}.issubset(df.columns):

    city_revenue = (
        df.groupby("City")["Revenue"]
        .sum()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=city_revenue.values,
        y=city_revenue.index
    )

    plt.title("Total Revenue by City", fontsize=16)
    plt.xlabel("Total Revenue")
    plt.ylabel("City")

    plt.tight_layout()
    plt.show()

# ============================================================
# 9. BAR CHART - ORDERS BY CUISINE
# ============================================================

if {"Cuisine", "Orders"}.issubset(df.columns):

    cuisine_orders = (
        df.groupby("Cuisine")["Orders"]
        .sum()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=cuisine_orders.values,
        y=cuisine_orders.index
    )

    plt.title("Total Orders by Cuisine", fontsize=16)
    plt.xlabel("Total Orders")
    plt.ylabel("Cuisine")

    plt.tight_layout()
    plt.show()

  # ============================================================
# 10. LINE PLOT - ORDERS OVER TIME
# ============================================================

if {"Date", "Orders"}.issubset(df.columns):

    daily_orders = (
        df.groupby("Date")["Orders"]
        .sum()
        .reset_index()
    )

    plt.figure(figsize=(12, 6))

    sns.lineplot(
        data=daily_orders,
        x="Date",
        y="Orders",
        marker="o"
    )

    plt.title("Food Delivery Orders Over Time", fontsize=16)
    plt.xlabel("Date")
    plt.ylabel("Number of Orders")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

  # ============================================================
# 11. LINE PLOT - REVENUE OVER TIME
# ============================================================

if {"Date", "Revenue"}.issubset(df.columns):

    daily_revenue = (
        df.groupby("Date")["Revenue"]
        .sum()
        .reset_index()
    )

    plt.figure(figsize=(12, 6))

    sns.lineplot(
        data=daily_revenue,
        x="Date",
        y="Revenue",
        marker="o"
    )

    plt.title("Food Delivery Revenue Over Time", fontsize=16)
    plt.xlabel("Date")
    plt.ylabel("Revenue")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ============================================================
# 12. BOX PLOT - REVENUE BY CITY
# ============================================================

if {"City", "Revenue"}.issubset(df.columns):

    plt.figure(figsize=(12, 6))

    sns.boxplot(
        data=df,
        x="City",
        y="Revenue"
    )

    plt.title("Revenue Distribution Across Cities", fontsize=16)
    plt.xlabel("City")
    plt.ylabel("Revenue")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ============================================================
# 13. VIOLIN PLOT - DELIVERY TIME BY CITY
# ============================================================

if {"City", "Delivery_Time"}.issubset(df.columns):

    plt.figure(figsize=(12, 6))

    sns.violinplot(
        data=df,
        x="City",
        y="Delivery_Time"
    )

    plt.title("Delivery Time Distribution Across Cities", fontsize=16)
    plt.xlabel("City")
    plt.ylabel("Delivery Time")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ============================================================
# 14. SCATTER PLOT - MARKETING SPEND VS REVENUE
# ============================================================

if {"Marketing_Spend", "Revenue"}.issubset(df.columns):

    plt.figure(figsize=(10, 6))

    sns.scatterplot(
        data=df,
        x="Marketing_Spend",
        y="Revenue"
    )

    plt.title("Marketing Spend vs Revenue", fontsize=16)
    plt.xlabel("Marketing Spend")
    plt.ylabel("Revenue")

    plt.tight_layout()
    plt.show()

# ============================================================
# 15. MARKETING SPEND VS REVENUE WITH TREND LINE
# ============================================================

if {"Marketing_Spend", "Revenue"}.issubset(df.columns):

    plt.figure(figsize=(10, 6))

    sns.regplot(
        data=df,
        x="Marketing_Spend",
        y="Revenue",
        scatter_kws={"alpha": 0.6}
    )

    plt.title("Relationship Between Marketing Spend and Revenue", fontsize=16)
    plt.xlabel("Marketing Spend")
    plt.ylabel("Revenue")

    plt.tight_layout()
    plt.show()

# ============================================================
# 17. WEATHER VS ORDERS
# ============================================================

if {"Weather", "Orders"}.issubset(df.columns):

    weather_orders = (
        df.groupby("Weather")["Orders"]
        .mean()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=weather_orders.index,
        y=weather_orders.values
    )

    plt.title("Average Orders by Weather Condition", fontsize=16)
    plt.xlabel("Weather Condition")
    plt.ylabel("Average Orders")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ============================================================
# 18. WEATHER VS REVENUE
# ============================================================

if {"Weather", "Revenue"}.issubset(df.columns):

    weather_revenue = (
        df.groupby("Weather")["Revenue"]
        .mean()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=weather_revenue.index,
        y=weather_revenue.values
    )

    plt.title("Average Revenue by Weather Condition", fontsize=16)
    plt.xlabel("Weather Condition")
    plt.ylabel("Average Revenue")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ============================================================
# 19. ORDER CHANNEL ANALYSIS
# ============================================================

if {"Order_Channel", "Orders"}.issubset(df.columns):

    channel_orders = (
        df.groupby("Order_Channel")["Orders"]
        .sum()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=channel_orders.index,
        y=channel_orders.values
    )

    plt.title("Total Orders by Order Channel", fontsize=16)
    plt.xlabel("Order Channel")
    plt.ylabel("Total Orders")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ============================================================
# 20. ORDER CHANNEL VS REVENUE
# ============================================================

if {"Order_Channel", "Revenue"}.issubset(df.columns):

    channel_revenue = (
        df.groupby("Order_Channel")["Revenue"]
        .sum()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=channel_revenue.index,
        y=channel_revenue.values
    )

    plt.title("Total Revenue by Order Channel", fontsize=16)
    plt.xlabel("Order Channel")
    plt.ylabel("Total Revenue")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ============================================================
# 21. CORRELATION HEATMAP
# ============================================================

numeric_data = df.select_dtypes(include=np.number)

if numeric_data.shape[1] >= 2:

    correlation = numeric_data.corr()

    plt.figure(figsize=(12, 8))

    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        linewidths=0.5
    )

    plt.title("Correlation Heatmap of Numerical Variables", fontsize=16)

    plt.tight_layout()
    plt.show()

# ============================================================
# 22. STRONGEST CORRELATIONS
# ============================================================

if numeric_data.shape[1] >= 2:

    correlation = numeric_data.corr()

    correlation_pairs = (
        correlation.where(
            np.triu(np.ones(correlation.shape), k=1).astype(bool)
        )
        .stack()
        .sort_values(key=abs, ascending=False)
    )

    print("Strongest numerical relationships:")
    display(correlation_pairs.head(10))

# ============================================================
# 23. MONTHLY ORDERS AND REVENUE
# ============================================================

if "Date" in df.columns:

    df["Month"] = df["Date"].dt.to_period("M").astype(str)

    monthly = (
        df.groupby("Month")
        .agg(
            Orders=("Orders", "sum") if "Orders" in df.columns else ("Month", "size"),
            Revenue=("Revenue", "sum") if "Revenue" in df.columns else ("Month", "size")
        )
        .reset_index()
    )

    print("Monthly summary:")
    display(monthly)

# ============================================================
# 24. MONTHLY ORDERS
# ============================================================

if {"Month", "Orders"}.issubset(df.columns):

    monthly_orders = df.groupby("Month")["Orders"].sum()

    plt.figure(figsize=(12, 6))

    sns.lineplot(
        x=monthly_orders.index,
        y=monthly_orders.values,
        marker="o"
    )

    plt.title("Monthly Order Trend", fontsize=16)
    plt.xlabel("Month")
    plt.ylabel("Total Orders")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
# ============================================================
# 25. MONTHLY REVENUE
# ============================================================

if {"Month", "Revenue"}.issubset(df.columns):

    monthly_revenue = df.groupby("Month")["Revenue"].sum()

    plt.figure(figsize=(12, 6))

    sns.lineplot(
        x=monthly_revenue.index,
        y=monthly_revenue.values,
        marker="o"
    )

    plt.title("Monthly Revenue Trend", fontsize=16)
    plt.xlabel("Month")
    plt.ylabel("Total Revenue")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ============================================================
# 26. OVERALL DATA SUMMARY
# ============================================================

print("=" * 60)
print("FOOD DELIVERY DATA VISUALIZATION - KEY FINDINGS")
print("=" * 60)

if "Orders" in df.columns:
    print("\n1. Total Orders:", df["Orders"].sum())
    print("   Average Orders:", round(df["Orders"].mean(), 2))

if "Revenue" in df.columns:
    print("\n2. Total Revenue:", round(df["Revenue"].sum(), 2))
    print("   Average Revenue:", round(df["Revenue"].mean(), 2))

if {"City", "Revenue"}.issubset(df.columns):
    best_city = df.groupby("City")["Revenue"].sum().idxmax()
    print("\n3. Highest Revenue City:", best_city)

if {"Cuisine", "Orders"}.issubset(df.columns):
    best_cuisine = df.groupby("Cuisine")["Orders"].sum().idxmax()
    print("4. Highest Order Cuisine:", best_cuisine)

if {"Marketing_Spend", "Revenue"}.issubset(df.columns):
    marketing_corr = df["Marketing_Spend"].corr(df["Revenue"])
    print("5. Marketing Spend vs Revenue Correlation:",
          round(marketing_corr, 2))

if {"Delivery_Time", "Customer_Rating"}.issubset(df.columns):
    delivery_corr = df["Delivery_Time"].corr(df["Customer_Rating"])
    print("6. Delivery Time vs Customer Rating Correlation:",
          round(delivery_corr, 2))

print("\n" + "=" * 60)
```
